```python
from crewai import LLM
llm = LLM(model="gemini/gemini-2.0-flash",
          temperature=0.5,
)
```

```python
from crewai import Agent, Task, Crew

email_assistant = Agent(role="Professional Email Assistant Agent",
                        goal="Improves messages to be more clear, concise, and professional.",
                        backstory="A highly experiened communication expect skilled in professional message writing",
                        verbose=True,
                        llm=llm
                        )
orginal_email = """Hey guys, i am conducting a master class on AI, GEn AI Stuff, you can join if you want, its free, just register using the link"""

email_Task = Task(description=f"""Take the following rough message and rewrite it into professional and polished version.
                  Expand abbrivation:
                  '''{orginal_email}'''""",\
                    agent=email_assistant,
                    expected_output="A professional written text message with proper formatting and content.")

crew = Crew(agents=[email_assistant],
            tasks=[email_Task],
            verbose=True)

result = crew.kickoff()  
```

```python
from crewai import Agent, Task, Crew

# Example raw code snippet
raw_code = """
import math
import logging
from typing import List

logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")

class PatternAnalyzer:
    def __init__(self, value: int):
        self.value = value
        self.components = self._decompose()
        self.power = len(self.components)

    def _decompose(self) -> List[int]:
        return [int(d) for d in str(self.value)]

    def _evaluate(self) -> float:
        return sum(math.pow(d, self.power) for d in self.components)

    def matches_rule(self) -> bool:
        total = self._evaluate()
        logging.debug(f"Analyzing {self.value}: computed {total}")
        return int(total) == self.value

    @staticmethod
    def scan_range(start: int, end: int) -> List[int]:
        results = []
        for num in range(start, end + 1):
            analyzer = PatternAnalyzer(num)
            if analyzer.matches_rule():
                results.append(num)
                logging.info(f"Special match found: {num}")
        return results


if __name__ == "__main__":
    value = 9474
    analyzer = PatternAnalyzer(value)
    print(f"{value} satisfies condition? {analyzer.matches_rule()}")
    start, end = 1, 100000
    matches = PatternAnalyzer.scan_range(start, end)
    print(f"Special matches between {start} and {end}:")
    print(matches)

"""
code_assistant = Agent(role="Purpose Finder Code Assistant Agent",
                        goal=" what for this code is used ",
                        backstory="A highly experienced code expert skilled in professional code writing",
                        verbose=True,
                        llm=llm
                        )

code_Task = Task(description=f"""Take the following code and gives what its used for.
                  Expand abbrivation:
                  '''{raw_code}'''""",
                    agent=code_assistant,
                    expected_output="Purpose of the code.")

crew = Crew(agents=[code_assistant],
            tasks=[code_Task],
            verbose=True)

result = crew.kickoff()  
```

```python
from crewai import Agent, Task, Crew
llm = LLM(model="gemini/gemini-2.0-flash",
          temperature=0.5,
          )
# Example raw code snippet
repo = "https://github.com/CTurt/FreeDVDBoot.git"
repo_scanner = Agent(role="Repository Scanner Agent",
                        goal="Scan the repository and parse the dependencies of the files",
                        backstory="A highly experienced Repository User",
                        verbose=True,
                        llm=llm
                        )

repo_Task = Task(
  description=f"""Repository Scanner:
                  Expand abbrivation:
                  '''{repo}'''""",
                    agent=repo_scanner,
                    expected_output="Scan the repository and parse the dependencies of the files.")

crew = Crew(
  agents=[repo_scanner],
  tasks=[repo_Task],
  verbose=True
)


result = crew.kickoff()  
```
